# Tape-width calibration previews

Generate sanitized preview labels for every configured tape width. This notebook does not connect to a printer. Widths other than 6, 9, and 18 mm are experimental and may require calibration.

```python
from IPython.display import display

from core.label_generator import LabelType, generate_label
from core.printing import get_supported_tape_sizes
```

```python
previews = {}

for tape_width_mm in get_supported_tape_sizes():
    label = generate_label(
        text=f"TEST-{tape_width_mm:02d}MM",
        label_type=LabelType.TEXT_QR,
        tape_size=tape_width_mm,
    )[0]
    previews[tape_width_mm] = label
    print(f"{tape_width_mm} mm: {label.size}")
    display(label)
```

To perform a physical test, use the queued REST endpoint described in `docs/API_EXAMPLES.md`. Keeping printer addresses out of the notebook prevents accidental disclosure through source or saved outputs.
